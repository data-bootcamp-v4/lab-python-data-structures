```python
# Step 4: Create an empty set for customer orders
customer_orders = set()
```

```python
# Step 5: Ask user to input 3 products for the order
print("\nEnter 3 products for the customer order:")
while len(customer_orders) < 3:
    order = input(f"Product {len(customer_orders)+1}: ").lower()
    if order not in products:
        print("Invalid product. Choose from the available products.")
    elif order in customer_orders:
        print("You already added this product.")
    else:
        customer_orders.add(order)
```

```python
# Step 6: Print the products in the customer_orders set
print("\nCustomer Orders:")
for item in customer_orders:
    print(item)
```

```python
# Step 7: Calculate order statistics
total_products_ordered = len(customer_orders)
percentage_ordered = (total_products_ordered / len(products)) * 100
order_status = (total_products_ordered, percentage_ordered)
```

```python
# Step 8: Print order statistics
print("\nOrder Statistics:")
print(f"Total Products Ordered: {order_status[0]}")
print(f"Percentage of Products Ordered: {order_status[1]:.2f}%")
```

```python
# Step 9: Update the inventory (subtract 1 for each ordered product)
for item in customer_orders:
    if inventory[item] > 0:
        inventory[item] -= 1
```

```python

# Step 10: Print updated inventory
print("\nUpdated Inventory:")
for product, qty in inventory.items():
    print(f"{product}: {qty}")
```
