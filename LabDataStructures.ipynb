```python
products = ["t-shirt", "mug", "hat", "book", "keychain"]

inventory = {}

for items in products: 
    quantity = int(input(f"Enter the {items} quantity here: "))
    inventory[items] = quantity

print(inventory)

### inventory["t-shirt"] = int(input("Enter quantity for t-shirt: "))
### inventory["mug"] = int(input("Enter quantity for mug: "))
### inventory["hat"] = int(input("Enter quantity for hat: "))
### inventory["book"] = int(input("Enter quantity for book: "))
### inventory["keychain"] = int(input("Enter quantity for keychain: "))

### the above is an example of taking the inventory without using for loops
```

```python
customer_orders = set()

for i in range(3):
    product_name = input(f"Enter the name of product {i + 1}: ")
    customer_orders.add(product_name)

print(customer_orders)


### product1 = input("Enter product 1: ")
### product2 = input("Enter product 2: ")
### product3 = input("Enter product 3: ")

### customer_orders.add(product1)
### customer_orders.add(product2)
### customer_orders.add(product3)

### again, the idea of inputting the data without using for loops
```

```python
total_products_ordered = len(customer_orders)
print(total_products_ordered)
```

```python
percent_ordered = (total_products_ordered / len(products)) * 100
print(percent_ordered)
```

```python
order_status = (total_products_ordered, percent_ordered) 

### first tuple I think I've used? 
```
