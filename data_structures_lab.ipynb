```python
# defining List
Products = [ "T-shirt","Mug","Hat","Book","Keychain"]
```

```python
# Creating empty Dictionary
Inventory = {}
```

```python
# input for each product quantity
for product in Products:
    quantity = int(input(f"Enter the quantitive of {product}: "))
    Inventory[product] = quantity
    print(f"{product} quantity is {quantity}")
print(Inventory)    

```

```python
# creating empty set
customer_orders = set()
# input from the customer 
count = 0
print("T-shirt","Mug","Hat","Book","Keychain")
while count < 3:
   
    order = str(input("Enter the product you want to order: "))
    customer_orders.add(order)
    count +=1
print(customer_orders)
```

```python
# Calculate order statistics
total_products_ordered = len(customer_orders)
total_available_products = len(Products)
percentage_ordered = (total_products_ordered / total_available_products) * 100
order_status = (total_products_ordered , percentage_ordered)


print(order_status)



```

```python
print(type(order_status))
```

```python
# printing order statistics
print("Order statics:")
print(f"total products ordered : {total_products_ordered}")
print(f"Percentage of products ordered : {percentage_ordered}%")
```

```python
# update the inventory
for ele in customer_orders:
    if ele in Inventory:
        Inventory[ele] -= 1
print("Updated Inventory:") 
print(Inventory)       
```

```python
for product, quantity in Inventory.items():
    print(f"{product}: {quantity}")
```
