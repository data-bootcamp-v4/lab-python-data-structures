```python
# products
products = ["t-shirt", "mug", "hat", "book", "keychain"]

# empty inventory dict
inventory = {}

# input quantity
print("Input quantity for each product:")
for product in products:
    quantity = int(input(f"{product.capitalize()}: "))
    inventory[product] = quantity

# cuatomer orders
customer_orders = set()

#User inputs three product names
print("\nInput three products a customer wants to order:")
for i in range(3):
    while True:
        order = input(f"Product {i+1}: ").strip().lower()
        if order in products:
            customer_orders.add(order)
            break
        else:
            print("Product is not available. Please choose a different products.")

print("\nCustomer Ordered Products:")
for product in customer_orders:
    print(product.capitalize())

#Calculate order statistics
total_products_ordered = len(customer_orders)
total_products_available = len(products)
percentage_ordered = (total_products_ordered / total_products_available) * 100

order_status = (total_products_ordered, percentage_ordered)



print("\nOrder Statistics:")
print(f"Total Products Ordered: {order_status[0]}")
print(f"Percentage of Products Ordered: {order_status[1]:.2f}%")

# Update inventory by subtracting 1 from each ordered product
for product in customer_orders:
    if inventory[product] > 0:
        inventory[product] -= 1

# Print updated inventory
print("\nUpdated Inventory:")
for product, quantity in inventory.items():
    print(f"{product.capitalize()}: {quantity}")
```
