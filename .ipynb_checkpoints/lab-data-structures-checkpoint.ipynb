```python
products= ["t-shirt", "mug", "hat", "book", "keychain"]
```

```python
print(products)
```

```python
type(products)
```

```python
inventory={}
```

```python
print("Please enter the quantity for each product:")
for product in products:
    quantity = int(input(f"Quantity of {product}: "))
    inventory[product] = quantity
```

```python
customer_orders=set()
```

```python
print("Please enter three products the customer wants to order:")
for i in range(3):
    order = input(f"Product {i+1}: ").strip().lower()
    while order not in products:
        print("Invalid product, please choose from:", products)
        order = input(f"Product {i+1}: ").strip().lower()
    
    customer_orders.add(order)
```

```python
print(customer_orders)
```

```python
total_products_ordered = len(customer_orders)
```

```python
percentage_ordered = (total_products_ordered / len(products)) * 100
```

```python
order_status = (total_products_ordered, percentage_ordered)
```

```python
print(f"Total Products Ordered: {order_status[0]}")
print(f"Percentage of Products Ordered: {order_status[1]:.2f}%")
```

```python
for product in customer_orders:
    inventory[product] -= 1
```

```python
for product, quantity in inventory.items():
    print(f"{product}: {quantity}")
```
